## Face and Facial Keypoint detection

After you've trained a neural network to detect facial keypoints, you can then apply this network to *any* image that includes faces. The neural network expects a Tensor of a certain size as input and, so, to detect any face, you'll first have to do some pre-processing.

1. Detect all the faces in an image using a face detector (we'll be using a Haar Cascade detector in this notebook).
2. Pre-process those face images so that they are grayscale, and transformed to a Tensor of the input size that your net expects. This step will be similar to the `data_transform` you created and applied in Notebook 2, whose job was tp rescale, normalize, and turn any iimage into a Tensor to be accepted as input to your CNN.
3. Use your trained model to detect facial keypoints on the image.

## Keypoint detection

Now, we'll loop over each detected face in an image (again!) only this time, you'll transform those faces in Tensors that your CNN can accept as input images.

### TODO: Transform each detected face into an input Tensor

You'll need to perform the following steps for each detected face:
1. Convert the face from RGB to grayscale
2. Normalize the grayscale image so that its color range falls in [0,1] instead of [0,255]
3. Rescale the detected face to be the expected square size for your CNN (224x224, suggested)
4. Reshape the numpy image into a torch image.

You may find it useful to consult to transformation code in `data_load.py` to help you perform these processing steps.


### TODO: Detect and display the predicted keypoints

After each face has been appropriately converted into an input Tensor for your network to see as input, you'll wrap that Tensor in a Variable() and can apply your `net` to each face. The ouput should be the predicted the facial keypoints. These keypoints will need to be "un-normalized" for display, and you may find it helpful to write a helper function like `show_keypoints`. You should end up with an image like the following with facial keypoints that closely match the facial features on each individual face:

<img src='https://github.com/pradhapmoorthi/P1_Facial_Keypoints/blob/master/.ipynb_checkpoints/images/michelle_detected.png?raw=1' width=30% height=30%/>




```python

import numpy as np
import cv2
import torch

CANVAS = 224  # matches training: RandomCrop(224)

def make_input_tensor_from_face(image_bgr, x, y, w, h):
    """
    TODO 1–4: Convert detected face into a CNN-ready input tensor.
    1) Convert RGB/BGR face to grayscale
    2) Normalize grayscale to [0,1]
    3) Rescale face to 224x224 (suggested size)
    4) Reshape numpy image -> torch image (C,H,W) with batch dim (1,1,224,224)
GR -> GRAY  (OpenCV images are BGR)    """
    roi_gray = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)

    # 2) Normalize to [0,1]
    roi_norm = roi_gray.astype(np.float32) / 255.0

    # 3) Rescale to 224x224
    roi_resized = cv2.resize(roi_norm, (CANVAS, CANVAS), interpolation=cv2.INTER_AREA)

    # 4) Reshape numpy -> torch: (H,W) -> (1,1,H,W)
    roi_tensor = torch.from_numpy(roi_resized).unsqueeze(0).unsqueeze(0)

    return roi_tensor  # shape: (1,1,224,224)


def predict_keypoints_for_face(net, image_bgr, x, y, w, h):
    """
    Runs the model on one face and returns keypoints mapped back to original image coords.
    Assumes training normalization: (kp - 100)/50 → inverse: kp = raw*50 + 100 (on 224x224 canvas).
    """
    net.eval()
    with torch.no_grad():
        inp = make_input_tensor_from_face(image_bgr, x, y, w, h)   # (1,1,224,224)
        raw = net(inp).view(-1, 2).cpu().numpy()                   # (68,2) standardized

    # inverse of training normalization
    pred_on_canvas = np.empty_like(raw, dtype=np.float32)
    pred_on_canvas[:, 0] = raw[:, 0] * 50.0 + 100.0   # x on 224 canvas
    pred_on_canvas[:, 1] = raw[:, 1] * 50.0 + 100.0   # y on 224 canvas

    # scale to the original face box and translate to image coordinates
    sx, sy = w / float(CANVAS), h / float(CANVAS)
    kp_xy = np.empty_like(pred_on_canvas, dtype=np.float32)
    kp_xy[:, 0] = x + pred_on_canvas[:, 0] * sx
    kp_xy[:, 1] = y + pred_on_canvas[:, 1] * sy

    return kp_xy  # shape: (68,2), in original image coordinates (pixel units)
    # crop the face ROI from original BGR image
    roi_bgr = image_bgr[y:y+h, x:x+w]


# faces is the array from detectMultiScale: [(x, y, w, h), ...]
all_keypoints = []
for (x, y, w, h) in faces:
    kp_xy = predict_keypoints_for_face(net, image_bgr, x, y, w, h)
    all_keypoints.append(kp_xy)
```

```python

import numpy as np
import cv2
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

def show_full_image_overlay_single_axes(image_bgr, faces, all_keypoints, fig_width=10.0):
    #for Matplotlib  Clean slate: close any prior figures that might auto-render afterward
    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
    H, W = image_rgb.shape[:2]
    aspect = W / float(H)

    # 2) Create ONE figure with no margins and a single Axes that fills it
    fig = plt.figure(figsize=(fig_width, fig_width / aspect), dpi=120)
    ax = fig.add_axes([0, 0, 1, 1])  # [left, bottom, width, height] in figure coords → full bleed
    ax.imshow(image_rgb)
    ax.set_axis_off()

    # 3) Lock axes to image extent (no padding); keep correct orientation for imshow
    ax.set_xlim(0, W)
    ax.set_ylim(H, 0)

    # 4) Draw boxes & keypoints on the SAME axes
    for (x, y, w, h), kp_xy in zip(faces, all_keypoints):
        # rectangle (cyan)
        ax.add_line(Line2D(
            [x, x+w, x+w, x,   x],
            [y, y,   y+h, y+h, y],
            color='#00FFFF', linewidth=2, zorder=2
        ))

        # keypoints (magenta), clipped to image bounds for neatness
        kx = np.clip(kp_xy[:, 0], 0, W-1)
        ky = np.clip(kp_xy[:, 1], 0, H-1)
        ax.scatter(kx, ky, s=22, c='#FF00FF', marker='o',
                   edgecolors='white', linewidths=0.5, zorder=3, clip_on=True)

    # 5) Render exactly once
    fig.canvas.draw_idle()
    plt.show()

# CALL IT (ensure the last line of the cell is the function call)
show_full_image_overlay_single_axes(image_bgr, faces, all_keypoints);
plt.close('all')

```

```python

import numpy as np
import cv2
import matplotlib.pyplot as plt

def show_rubric_compliant_overlay(image_bgr, faces, all_keypoints,
                                  box_color='#00FFFF',  # cyan
                                  kp_color='#FF00FF',   # magenta
                                  marker_size=22,
                                  fig_width=10.0):
    """
    image_bgr: np.ndarray (H,W,3) from cv2.imread
    faces: list/array of (x, y, w, h) from detectMultiScale
    all_keypoints: list of (68,2) arrays in original image coordinates
    """

    # clean slate for notebooks
    plt.close('all')

    # convert BGR->RGB for matplotlib and compute aspect
    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
    H, W = image_rgb.shape[:2]
    aspect = W / float(H)

    # single figure, single axes, full-bleed (no margins)
    fig = plt.figure(figsize=(fig_width, fig_width / aspect), dpi=120)
    ax = fig.add_axes([0, 0, 1, 1])  # full canvas
    ax.imshow(image_rgb)
    ax.set_axis_off()

    # lock axes to image extent (no padding) and keep orientation
    ax.set_xlim(0, W)
    ax.set_ylim(H, 0)

    # draw boxes & keypoints on the SAME axes
    for (x, y, w, h), kp_xy in zip(faces, all_keypoints):
        # rectangle
        ax.plot([x, x+w, x+w, x, x],
                [y, y,   y+h, y+h, y],
                color=box_color, linewidth=2, zorder=2)

        # keypoints (clip to image bounds for cleanliness)
        kx = np.clip(kp_xy[:, 0], 0, W-1)
        ky = np.clip(kp_xy[:, 1], 0, H-1)
        ax.scatter(kx, ky, s=marker_size, c=kp_color, marker='o',
                   edgecolors='white', linewidths=0.5, zorder=3, clip_on=True)

    # render once, with zero padding
    fig.canvas.draw_idle()
    plt.show()
```

```python
show_rubric_compliant_overlay(image_bgr, faces, all_keypoints)
```
