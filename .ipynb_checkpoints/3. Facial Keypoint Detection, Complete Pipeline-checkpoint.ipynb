## Face and Facial Keypoint detection

After you've trained a neural network to detect facial keypoints, you can then apply this network to *any* image that includes faces. The neural network expects a Tensor of a certain size as input and, so, to detect any face, you'll first have to do some pre-processing.

1. Detect all the faces in an image using a face detector (we'll be using a Haar Cascade detector in this notebook).
2. Pre-process those face images so that they are grayscale, and transformed to a Tensor of the input size that your net expects. This step will be similar to the `data_transform` you created and applied in Notebook 2, whose job was tp rescale, normalize, and turn any iimage into a Tensor to be accepted as input to your CNN.
3. Use your trained model to detect facial keypoints on the image.

```python
# load in a haar cascade classifier for detecting frontal faces
face_cascade = cv2.CascadeClassifier('/content/gdrive/MyDrive/P1_Facial_Keypoints-master/detector_architectures/haarcascade_frontalface_default.xml')

# run the detector
# the output here is an array of detections; the corners of each detection box
# if necessary, modify these parameters until you successfully identify every face in a given image
faces = face_cascade.detectMultiScale(image, 1.2, 2)

# make a copy of the original image to plot detections on
image_with_detections = image.copy()

# loop over the detected faces, mark the image where each face is found
for (x,y,w,h) in faces:
    # draw a rectangle around each detected face
    # you may also need to change the width of the rectangle drawn depending on image resolution
    cv2.rectangle(image_with_detections,(x,y),(x+w,y+h),(255,0,0),3)

fig = plt.figure(figsize=(9,9))

plt.imshow(image_with_detections)
```

## Keypoint detection

Now, we'll loop over each detected face in an image (again!) only this time, you'll transform those faces in Tensors that your CNN can accept as input images.

### TODO: Transform each detected face into an input Tensor

You'll need to perform the following steps for each detected face:
1. Convert the face from RGB to grayscale
2. Normalize the grayscale image so that its color range falls in [0,1] instead of [0,255]
3. Rescale the detected face to be the expected square size for your CNN (224x224, suggested)
4. Reshape the numpy image into a torch image.

You may find it useful to consult to transformation code in `data_load.py` to help you perform these processing steps.


### TODO: Detect and display the predicted keypoints

After each face has been appropriately converted into an input Tensor for your network to see as input, you'll wrap that Tensor in a Variable() and can apply your `net` to each face. The ouput should be the predicted the facial keypoints. These keypoints will need to be "un-normalized" for display, and you may find it helpful to write a helper function like `show_keypoints`. You should end up with an image like the following with facial keypoints that closely match the facial features on each individual face:

<img src='https://github.com/pradhapmoorthi/P1_Facial_Keypoints/blob/master/.ipynb_checkpoints/images/michelle_detected.png?raw=1' width=30% height=30%/>




```python
# Defining a function for plotting keypoints on the images
def show_all_keypoints(image, predicted_key_pts):
    """Show image with predicted keypoints"""
    # image is grayscale
    plt.imshow(image, cmap='gray')
    plt.scatter(predicted_key_pts[:, 0], predicted_key_pts[:, 1], s=20, marker='.', c='m')
```

```python
import numpy as np

image_copy = np.copy(image)

# loop over the detected faces from your haar cascade
for (x,y,w,h) in faces:

    # Select the region of interest that is the face in the image
    # The sizes of faces detected by a Haar detector and the faces our network has been trained on are of different sizes. Hence, padding the input image.
    padding= 50                                          ## padding the image (adjust this parameter, if required)

    # Calculate coordinates with padding, ensuring they stay within image bounds
    y1 = max(0, y - padding)
    y2 = min(image_copy.shape[0], y + h + padding)
    x1 = max(0, x - padding)
    x2 = min(image_copy.shape[1], x + w + padding)

    roi = image_copy[y1:y2, x1:x2]

    # Check if ROI is empty after clipping (e.g., if a tiny face is too close to border)
    if roi.shape[0] == 0 or roi.shape[1] == 0:
        print(f"Skipping empty ROI for face at ({x},{y},{w},{h})")
        continue

    ## TODO: Convert the face region from RGB to grayscale
    roi_gray = cv2.cvtColor(roi, cv2.COLOR_RGB2GRAY)

    ## TODO: Normalize the grayscale image so that its color range falls in [0,1] instead of [0,255]
    # scale color range from [0, 255] to [0, 1]
    roi_norm_img=  roi_gray/255.0

    ## TODO: Rescale the detected face to be the expected square size for your CNN (224x224, suggested)
    new_size= 224
    roi_resize= cv2.resize(roi_norm_img, (new_size, new_size))   ## Note: new_size= (WxH)
    roi_resize_copy= np.copy(roi_resize)  # Create a copy for later use
    '''IMPORTANT: 'roi_resize_copy' will be later used as the base image on which resized keypoints will be plotted.'''

    ## TODO: Reshape the numpy image shape (H x W x C) into a torch image shape (C x H x W)
    roi_reshape= roi_resize.reshape(1, roi_resize.shape[1], roi_resize.shape[0]) # Numpy (C x H x W)
    roi_tensor= torch.Tensor(roi_reshape)  ## Convert into torch image (tensor) of size (C X H X W)
    roi_tensor = roi_tensor.unsqueeze(0)  ## Proper Input size:Refer to the function 'net_sample_output()'' code in Notebook 2 (For understanding)

    ## TODO: Make facial keypoint predictions using your loaded, trained network
    roi_tensor = roi_tensor.type(torch.FloatTensor)     # Convert images to FloatTensors
    ## forward pass to get net output (net.eval(): Refer to the network cell above. The program is already in evaluation mode)
    with torch.no_grad():
        output_pts = net(roi_tensor)                    # Predict keypoints
    # keypoints_output: reshape to batch_size x 68 x 2 pts
    output_pts = output_pts.view(output_pts.size()[0], 68, -1)   # output_pts.size()[0]: 1 for 1 image input
    # un-transform the predicted key_pts data
    with torch.no_grad():
        predicted_key_pts = np.squeeze(output_pts.numpy())       # Convert from torch tensor to numpy()
    # undo normalization of keypoints
    predicted_key_pts = predicted_key_pts*50.0+100   ## For more info: Refer to class normalize in 'data_load.py' or class normalize in notebook 1


    ## TODO: Display each detected face and the corresponding keypoints
    show_all_keypoints(roi_resize_copy, predicted_key_pts)  ## Using the transformed image numpy copy (224,224): roi_resize_copy
    plt.axis('off')    # Display images without axes
    plt.show()
```
