## Face and Facial Keypoint detection

After you've trained a neural network to detect facial keypoints, you can then apply this network to *any* image that includes faces. The neural network expects a Tensor of a certain size as input and, so, to detect any face, you'll first have to do some pre-processing.

1. Detect all the faces in an image using a face detector (we'll be using a Haar Cascade detector in this notebook).
2. Pre-process those face images so that they are grayscale, and transformed to a Tensor of the input size that your net expects. This step will be similar to the `data_transform` you created and applied in Notebook 2, whose job was tp rescale, normalize, and turn any iimage into a Tensor to be accepted as input to your CNN.
3. Use your trained model to detect facial keypoints on the image.

**Reasoning**:
The next logical step is to load the image specified by `image_path` using OpenCV, convert it to RGB format for consistent display with matplotlib, and create a copy of the image for drawing detections.



```python
import cv2
import matplotlib.pyplot as plt

# Load the image
image = cv2.imread(image_path)

# Handle potential errors if the image cannot be loaded
if image is None:
    print(f"Error: Image not found or could not be loaded at {image_path}. Please check the file path.")
else:
    # Convert the image from BGR (OpenCV default) to RGB (matplotlib default)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    # Create a copy of the image to plot detections on later
    image_copy = np.copy(image)

    # Display the loaded image
    fig = plt.figure(figsize=(9,9))
    plt.imshow(image)
    plt.title(f"Loaded Image: {os.path.basename(image_path)}")
    plt.axis('off')
    plt.show()
    print(f"Image shape: {image.shape}")
```

**Reasoning**:
To perform face detection, I need to load the Haar Cascade classifier and then apply it to the `image` to identify faces. After detection, I will draw bounding boxes on a copy of the image to visualize the detected faces.



```python
import cv2

# Load the Haar Cascade classifier for frontal faces
face_cascade = cv2.CascadeClassifier('/content/drive/MyDrive/P1_Facial_Keypoints-master/detector_architectures/haarcascade_frontalface_default.xml')

# Detect faces in the image_copy (the RGB image loaded previously)
# Adjust parameters (scaleFactor, minNeighbors) if needed to improve detection
faces = face_cascade.detectMultiScale(image, 1.2, 2)

# Make a copy of the original image to plot detections on
image_with_detections = image.copy()

# Loop over the detected faces and draw a rectangle around each
for (x,y,w,h) in faces:
    cv2.rectangle(image_with_detections,(x,y),(x+w,y+h),(255,0,0),3) # Red rectangle with thickness 3

# Display the image with detected faces
plt.figure(figsize=(9,9))
plt.imshow(image_with_detections)
plt.title("Faces Detected")
plt.axis('off')
plt.show()

print(f"Found {len(faces)} face(s) in the image.")
```

**Reasoning**:
Now that faces are detected, the next step is to load the pre-trained facial keypoint detection model and set it to evaluation mode, which is necessary before making predictions.



```python
import torch
from models import Net

net = Net()

## Load the best saved model parameters
net.load_state_dict(torch.load('/content/drive/MyDrive/saved_models/keypoints_SimpleAugVar2_best.pt', map_location=torch.device('cpu')))

## Set the net to evaluation mode
net.eval()
print(net)
```

**Reasoning**:
Now that the model is loaded and faces are detected, the next step is to iterate through each detected face, perform the necessary pre-processing, predict keypoints, determine face orientation and smile status, and then visualize these results. A summary of findings for each face will also be collected.



```python
import numpy as np
import cv2
import torch
import matplotlib.pyplot as plt
import os

# Defining a function for plotting keypoints on the images
def show_all_keypoints(image, predicted_key_pts, orientation='', smile_status=''):
    """Show image with predicted keypoints, orientation, and smile status"""
    plt.imshow(image, cmap='gray')
    plt.scatter(predicted_key_pts[:, 0], predicted_key_pts[:, 1], s=20, marker='.', c='m')
    if orientation: # Only add text if orientation is not empty
        plt.text(5, 20, f'Orientation: {orientation}', color='cyan', fontsize=12, bbox=dict(facecolor='black', alpha=0.5))
    if smile_status: # Only add text if smile_status is not empty
        plt.text(5, 40, f'Smile: {smile_status}', color='cyan', fontsize=12, bbox=dict(facecolor='black', alpha=0.5))


results_summary = []

# Loop over the detected faces from your Haar cascade
for i, (x, y, w, h) in enumerate(faces):
    # Select the region of interest that is the face in the image
    # The sizes of faces detected by a Haar detector and the faces our network has been trained on are of different sizes. Hence, padding the input image.
    padding = 50                                          ## padding the image (adjust this parameter, if required)

    # Calculate coordinates with padding, ensuring they stay within image bounds
    y1 = max(0, y - padding)
    y2 = min(image.shape[0], y + h + padding)
    x1 = max(0, x - padding)
    x2 = min(image.shape[1], x + w + padding)

    roi = image[y1:y2, x1:x2]

    # Check if ROI is empty after clipping (e.g., if a tiny face is too close to border)
    if roi.shape[0] == 0 or roi.shape[1] == 0:
        print(f"Skipping empty ROI for face {i+1} at ({x},{y},{w},{h})")
        continue

    # Convert the face region from RGB to grayscale
    roi_gray = cv2.cvtColor(roi, cv2.COLOR_RGB2GRAY)

    # Normalize the grayscale image so that its color range falls in [0,1] instead of [0,255]
    roi_norm_img = roi_gray / 255.0

    # Rescale the detected face to be the expected square size for your CNN (224x224)
    new_size = 224
    roi_resize = cv2.resize(roi_norm_img, (new_size, new_size))
    roi_resize_copy = np.copy(roi_resize)  # Create a copy for later use

    # Reshape the numpy image shape (H x W) into a torch image shape (C x H x W)
    roi_reshape = roi_resize.reshape(1, roi_resize.shape[0], roi_resize.shape[1])
    roi_tensor = torch.from_numpy(roi_reshape)  ## Convert into torch image (tensor)
    roi_tensor = roi_tensor.unsqueeze(0)  ## Add batch dimension: 1 x C x H x W

    # Make facial keypoint predictions using your loaded, trained network
    roi_tensor = roi_tensor.type(torch.FloatTensor)     # Convert images to FloatTensors
    with torch.no_grad():
        output_pts = net(roi_tensor)                    # Predict keypoints

    # Reshape the output keypoints data
    output_pts = output_pts.view(output_pts.size()[0], 68, -1)
    predicted_key_pts = np.squeeze(output_pts.numpy())       # Convert from torch tensor to numpy()

    # Un-normalize the predicted keypoints
    predicted_key_pts = predicted_key_pts * 50.0 + 100   ## For more info: Refer to class normalize in 'data_load.py' or class normalize in notebook 1

    # Determine Face Orientation
    nose_tip_x = predicted_key_pts[30, 0]
    face_center_x = new_size / 2
    threshold = new_size * 0.1 # A 10% deviation from center

    orientation = ''
    if nose_tip_x < (face_center_x - threshold):
        orientation = 'Right-looking' # Nose is to the left of center, so face is looking right
    elif nose_tip_x > (face_center_x + threshold):
        orientation = 'Left-looking'  # Nose is to the right of center, so face is looking left
    else:
        orientation = 'Center'

    # Detect Smile (using mouth keypoints)
    # Mouth keypoints: 48-67
    left_mouth_corner_x = predicted_key_pts[48, 0]
    right_mouth_corner_x = predicted_key_pts[54, 0]
    upper_lip_y = predicted_key_pts[51, 1]
    lower_lip_y = predicted_key_pts[57, 1]

    mouth_width = abs(right_mouth_corner_x - left_mouth_corner_x)
    mouth_height = abs(lower_lip_y - upper_lip_y)

    smile_ratio_threshold = 0.35 # This threshold might need fine-tuning
    smile_status = ''
    # A wider and relatively higher mouth often indicates a smile
    if mouth_width > 0 and (mouth_height / mouth_width) > smile_ratio_threshold:
        smile_status = 'Smiling'
    else:
        smile_status = 'Not Smiling'

    # Display each detected face and the corresponding keypoints, orientation, and smile status
    plt.figure(figsize=(6, 6))
    show_all_keypoints(roi_resize_copy, predicted_key_pts, orientation, smile_status)
    plt.title(f'Face {i+1} - Orientation: {orientation}, Smile: {smile_status}')
    plt.axis('off')    # Display images without axes
    plt.show()

    # Store results for summary
    results_summary.append({
        'image_file': os.path.basename(image_path),
        'face_bbox': (x, y, w, h),
        'orientation': orientation,
        'smile_status': smile_status
    })

# Print the overall summary after processing all faces
print("\n--- Analysis Summary for all faces in the image ---")
for res in results_summary:
    print(f"Face BBox: {res['face_bbox']}, Orientation: {res['orientation']}, Smile: {res['smile_status']}")
```
