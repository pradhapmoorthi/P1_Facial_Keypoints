## PyTorch Neural Nets

To define a neural network in PyTorch, you define the layers of a model in the function `__init__` and define the feedforward behavior of a network that employs those initialized layers in the function `forward`, which takes in an input image tensor, `x`. The structure of this Net class is shown below and left for you to fill in.

Note: During training, PyTorch will be able to perform backpropagation by keeping track of the network's feedforward behavior and using autograd to calculate the update to the weights in the network.

#### Define the Layers in ` __init__`
As a reminder, a conv/pool layer may be defined like this (in `__init__`):
```
# 1 input image channel (for grayscale images), 32 output channels/feature maps, 3x3 square convolution kernel
self.conv1 = nn.Conv2d(1, 32, 3)

# maxpool that uses a square window of kernel_size=2, stride=2
self.pool = nn.MaxPool2d(2, 2)      
```

#### Refer to Layers in `forward`
Then referred to in the `forward` function like this, in which the conv1 layer has a ReLu activation applied to it before maxpooling is applied:
```
x = self.pool(F.relu(self.conv1(x)))
```

Best practice is to place any layers whose weights will change during the training process in `__init__` and refer to them in the `forward` function; any layers or functions that always behave in the same way, such as a pre-defined activation function, should appear *only* in the `forward` function.

#### Load in the test dataset

The test dataset is one that this model has *not* seen before, meaning it has not trained with these images. We'll load in this test data and before and after training, see how your model performs on this set!

To visualize this test data, we have to go through some un-transformation steps to turn our images into python images from tensors and to turn our keypoints back into a recognizable range.

````python
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

# the dataset we created in Notebook 1 is copied in the helper file `data_load.py`
from data_load import FacialKeypointsDataset
# the transforms we defined in Notebook 1 are in the helper file `data_load.py`
from data_load import Rescale, RandomCrop, Normalize, ToTensor, RandomRotation, RandomHorizontalFlip


# 1. Create an instance of FacialKeypointsDataset without transformations
raw_train_dataset = FacialKeypointsDataset(
    csv_file='data/training_frames_keypoints.csv',
    root_dir='data/training/'
)

print(f"Raw training dataset created with {len(raw_train_dataset)} samples.")

# 2. Instantiate RandomHorizontalFlip and RandomRotation transforms
flip_transform = RandomHorizontalFlip()
rotate_pos_10 = RandomRotation(degrees=10)
rotate_neg_10 = RandomRotation(degrees=-10)

print("Data augmentation transforms (Flip and Rotations) instantiated.")

# 3. Initialize an empty list called augmented_samples
augmented_samples = []

# Populate augmented_samples
# First, add all original samples
for i in range(len(raw_train_dataset)):
    augmented_samples.append(raw_train_dataset[i])

# Then, add flipped samples
for i in range(len(raw_train_dataset)):
    sample = raw_train_dataset[i]
    flipped_sample = flip_transform(sample)
    augmented_samples.append(flipped_sample)

# Next, add rotated samples for +/-15 and +/-30 degrees
for i in range(len(raw_train_dataset)):
    sample = raw_train_dataset[i]

    rotated_pos_10_sample = rotate_pos_10(sample)
    augmented_samples.append(rotated_pos_10_sample)

    rotated_neg_10_sample = rotate_neg_10(sample)
    augmented_samples.append(rotated_neg_10_sample)


augmented_dataset_size = len(augmented_samples)
# Original + Flip + 4 rotations (15_pos, 15_neg, 30_pos, 30_neg)
expected_augmented_size = len(raw_train_dataset) * 4

print(f"Number of augmented samples created: {augmented_dataset_size}")
print(f"Expected augmented dataset size: {expected_augmented_size}")
assert augmented_dataset_size == expected_augmented_size, "Augmented dataset size mismatch!"

print("Augmented samples successfully generated.")


# Define the CustomAugmentedDataset class
class CustomAugmentedDataset(Dataset):
    """Custom Dataset for pre-augmented facial keypoints samples."""

    def __init__(self, augmented_samples, transform=None):
        self.augmented_samples = augmented_samples
        self.transform = transform

    def __len__(self):
        return len(self.augmented_samples)

    def __getitem__(self, idx):
        sample = self.augmented_samples[idx]
        # The sample dictionary already contains 'image' (PIL Image) and 'keypoints' (numpy array)

        if self.transform:
            sample = self.transform(sample)

        return sample

# Define a final_data_transform
final_data_transform = transforms.Compose([
    Rescale(250),
    RandomCrop(224),
    RandomRotation(degrees=10),
    RandomHorizontalFlip(),
    Normalize(),
    ToTensor()
])

# Create an instance of CustomAugmentedDataset
augmented_train_dataset = CustomAugmentedDataset(
    augmented_samples=augmented_samples,
    transform=final_data_transform
)

print(f"CustomAugmentedDataset created with {len(augmented_train_dataset)} samples.")

# Create an augmented_train_loader
augmented_batch_size = 64

augmented_train_loader = DataLoader(augmented_train_dataset,
                                    batch_size=augmented_batch_size,
                                    shuffle=True,
                                    num_workers=2)

print(f"Augmented training DataLoader created with batch size {augmented_batch_size}.")
print(f"Number of batches in augmented_train_loader: {len(augmented_train_loader)}")
````

## Apply the model on a test sample

To test the model on a test sample of data, you have to follow these steps:
1. Extract the image and ground truth keypoints from a sample
2. Make sure the image is a FloatTensor, which the model expects.
3. Forward pass the image through the net to get the predicted, output keypoints.

This function test how the network performs on the first batch of test data. It returns the images, the transformed images, the predicted keypoints (produced by the model), and the ground truth keypoints.

````python
# test the model on a batch of test images
import torch

def net_sample_output():

    # iterate through the test dataset
    for i, sample in enumerate(test_loader):

        # get sample data: images and ground truth keypoints
        images = sample['image']
        key_pts = sample['keypoints']

        # convert images to FloatTensors
        images = images.type(torch.FloatTensor)

        # forward pass to get net output
        output_pts = net(images)

        # reshape to batch_size x 68 x 2 pts
        output_pts = output_pts.view(output_pts.size()[0], 68, -1)

        # break after first image is tested
        if i == 0:
            return images, output_pts, key_pts
````

## Visualize the predicted keypoints

Once we've had the model produce some predicted output keypoints, we can visualize these points in a way that's similar to how we've displayed this data before, only this time, we have to "un-transform" the image/keypoint data to display it.

Note that I've defined a *new* function, `show_all_keypoints` that displays a grayscale image, its predicted keypoints and its ground truth keypoints (if provided).

````python
def show_all_keypoints(image, predicted_key_pts, gt_pts=None):
    """Show image with predicted keypoints"""
    # image is grayscale
    plt.imshow(image, cmap='gray')
    plt.scatter(predicted_key_pts[:, 0], predicted_key_pts[:, 1], s=20, marker='.', c='m')
    # plot ground truth points as green pts
    if gt_pts is not None:
        plt.scatter(gt_pts[:, 0], gt_pts[:, 1], s=20, marker='.', c='g')
````

#### Un-transformation

Next, you'll see a helper function. `visualize_output` that takes in a batch of images, predicted keypoints, and ground truth keypoints and displays a set of those images and their true/predicted keypoints.

This function's main role is to take batches of image and keypoint data (the input and output of your CNN), and transform them into numpy images and un-normalized keypoints (x, y) for normal display. The un-transformation process turns keypoints and images into numpy arrays from Tensors *and* it undoes the keypoint normalization done in the Normalize() transform; it's assumed that you applied these transformations when you loaded your test data.

````python
# visualize the output
# by default this shows a batch of 10 images
def visualize_output(test_images, test_outputs, gt_pts=None, batch_size=10):

    for i in range(batch_size):
        plt.figure(figsize=(20,10))
        ax = plt.subplot(1, batch_size, i+1)

        # un-transform the image data
        image = test_images[i].data   # get the image from it's wrapper
        image = image.numpy()   # convert to numpy array from a Tensor
        image = np.transpose(image, (1, 2, 0))   # transpose to go from torch to numpy image

        # un-transform the predicted key_pts data
        predicted_key_pts = test_outputs[i].data
        predicted_key_pts = predicted_key_pts.numpy()
        # undo normalization of keypoints
        predicted_key_pts = predicted_key_pts*50.0+100

        # plot ground truth points for comparison, if they exist
        ground_truth_pts = None
        if gt_pts is not None:
            ground_truth_pts = gt_pts[i]
            ground_truth_pts = ground_truth_pts*50.0+100

        # call show_all_keypoints
        show_all_keypoints(np.squeeze(image), predicted_key_pts, ground_truth_pts)

        plt.axis('off')

    plt.show()

# call it
visualize_output(test_images=test_images, test_outputs=test_outputs, gt_pts=gt_pts)
````

## Training

#### Loss function
Training a network to predict keypoints is different than training a network to predict a class; instead of outputting a distribution of classes and using cross entropy loss, you may want to choose a loss function that is suited for regression, which directly compares a predicted value and target value. Read about the various kinds of loss functions (like MSE or L1/SmoothL1 loss) in [this documentation](https://pytorch.org/docs/master/nn.html#loss-functions).

### TODO: Define the loss and optimization

Next, you'll define how the model will train by deciding on the loss function and optimizer.

---

````python
## TODO: Define the loss and optimization
import torch.optim as optim
import torch.nn as nn # Added import for torch.nn

criterion = nn.SmoothL1Loss(beta=1.0)

optimizer = optim.Adam(net.parameters(), lr=1e-3, betas=(0.9, 0.999), weight_decay=1e-4)
````

````python
import torch.optim.lr_scheduler # Added import for lr_scheduler
import numpy as np # Import numpy for np.Inf
import time # Import the time module

def train_net(n_epochs, train_loader, test_loader):

    # initialize lists to store losses
    train_losses = []
    valid_losses = []

    # Initialize best validation loss and best model state
    best_valid_loss = np.inf # Corrected from np.Inf to np.inf
    best_model_state = None # To store the state_dict of the best model

    # prepare the net for training
    net.train()

    # Initialize the learning rate scheduler
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, factor=0.1, patience=10)

    for epoch in range(n_epochs):
        start_time = time.time() # Record start time of the epoch

        # Initialize variables for epoch-wise training loss calculation
        total_epoch_train_loss = 0.0
        num_train_batches = 0

        # Initialize variable for intermediate training loss printing
        running_loss_for_print = 0.0

        # training loop
        for batch_i, data in enumerate(train_loader):
            # get the input images and their corresponding labels
            images = data['image']
            key_pts = data['keypoints']

            # flatten pts
            key_pts = key_pts.view(key_pts.size(0), -1)

            # convert variables to floats for regression loss
            key_pts = key_pts.type(torch.FloatTensor)
            images = images.type(torch.FloatTensor)

            # forward pass to get outputs
            output_pts = net(images)

            # calculate the loss between predicted and target keypoints
            loss = criterion(output_pts, key_pts)

            # zero the parameter (weight) gradients
            optimizer.zero_grad()

            # backward pass to calculate the weight gradients
            loss.backward()

            # update the weights
            optimizer.step()

            # Accumulate loss for the entire epoch's average calculation
            total_epoch_train_loss += loss.item()
            num_train_batches += 1

            # Accumulate loss for intermediate printing and reset every 10 batches
            running_loss_for_print += loss.item()
            if batch_i % 64 == 63:
                print('Epoch: {}, Batch: {}, Avg. Training Loss (last 64 batches): {}'.format(epoch + 1, batch_i+1, running_loss_for_print/1000))
                running_loss_for_print = 0.0

        # Calculate and store average training loss for the entire epoch
        avg_train_loss = total_epoch_train_loss / num_train_batches
        train_losses.append(avg_train_loss)
        print('Epoch: {}, Total Avg. Training Loss: {}'.format(epoch + 1, avg_train_loss))

        # validation loop
        net.eval() # set model to evaluation mode
        valid_loss_sum = 0.0
        num_valid_batches = 0
        with torch.no_grad(): # disable gradient calculations
            for batch_i, data in enumerate(test_loader):
                images = data['image']
                key_pts = data['keypoints']

                key_pts = key_pts.view(key_pts.size(0), -1)

                key_pts = key_pts.type(torch.FloatTensor)
                images = images.type(torch.FloatTensor)

                output_pts = net(images)
                loss = criterion(output_pts, key_pts)
                valid_loss_sum += loss.item()
                num_valid_batches += 1

        # calculate average validation loss for the epoch
        avg_valid_loss = valid_loss_sum / num_valid_batches
        valid_losses.append(avg_valid_loss)
        print('Epoch: {}, Avg. Validation Loss: {}'.format(epoch + 1, avg_valid_loss))

        # Check if this is the best model so far
        if avg_valid_loss < best_valid_loss:
            best_valid_loss = avg_valid_loss
            best_model_state = net.state_dict() # Save the state_dict of the current best model
            print(f"--> New best model saved with validation loss: {best_valid_loss:.4f}")

        # Step the learning rate scheduler based on validation loss
        scheduler.step(avg_valid_loss)

        net.train() # set model back to training mode

        end_time = time.time() # Record end time of the epoch
        epoch_duration = end_time - start_time
        print(f'Epoch {epoch + 1} took {epoch_duration:.2f} seconds.') # Display epoch duration

    print('Finished Training')
    return train_losses, valid_losses, best_model_state, best_valid_loss # Return the best model state and its loss
````

## Feature Visualization

Sometimes, neural networks are thought of as a black box, given some input, they learn to produce some output. CNN's are actually learning to recognize a variety of spatial patterns and you can visualize what each convolutional layer has been trained to recognize by looking at the weights that make up each convolutional kernel and applying those one at a time to a sample image. This technique is called feature visualization and it's useful for understanding the inner workings of a CNN.

In the cell below, you can see how to extract a single filter (by index) from your first convolutional layer. The filter should appear as a grayscale grid.

````python
# Get the weights in the first conv layer, "conv1"
# if necessary, change this to reflect the name of your first conv layer
weights1 = net.conv1.weight.data

w = weights1.numpy()

filter_index = 0

print(w[filter_index][0])
print(w[filter_index][0].shape)

# display the filter weights
plt.imshow(w[filter_index][0], cmap='gray')
````

## Feature maps

Each CNN has at least one convolutional layer that is composed of stacked filters (also known as convolutional kernels). As a CNN trains, it learns what weights to include in its convolutional kernels and when these kernels are applied to some input image, they produce a set of **feature maps**. So, feature maps are just sets of filtered images; they are the images produced by applying a convolutional kernel to an input image. These maps show us the features that the different layers of the neural network learn to extract. For example, you might imagine a convolutional kernel that detects the vertical edges of a face or another one that detects the corners of eyes. ### Data Analysis Key Findings
*   Five unique random images and their ground truth keypoints were successfully selected from the `test_dataset`, which contains 770 samples.
*   The x-coordinate of the nose tip (keypoint 30) was extracted as the sole feature to represent facial orientation for each image.
*   K-means clustering was performed with $k=3$ to classify these orientations, corresponding to 'left-oriented', 'right-oriented', and 'centered' categories.

### Insights or Next Steps
*   The use of a single, simple feature like the nose tip's x-coordinate can be surprisingly effective for broad categorization of facial orientations, but may lack the granularity for finer distinctions.
*   For more robust and detailed facial orientation classification, consider exploring multi-dimensional features by incorporating additional keypoints (e.g., eye corners, jawline points) or applying dimensionality reduction techniques like PCA on a wider set of keypoints.


````python
import numpy as np
from sklearn.cluster import KMeans

# Un-transformation function (copied from visualize_output in the notebook context)
def un_transform_keypoints(key_pts_tensor):
    key_pts_np = key_pts_tensor.numpy()
    # undo normalization of keypoints
    key_pts_unnorm = key_pts_np * 50.0 + 100
    return key_pts_unnorm

# Prepare features for clustering
# For simplicity, let's use the average x-coordinate of the left side of the face vs. the right side
# Based on the facial keypoints diagram, keypoints 0-16 are jawline, 17-21 left eyebrow, 22-26 right eyebrow
# 36-41 left eye, 42-47 right eye, 48-59 mouth outer, 60-67 mouth inner
# Let's use the average x-coordinate of left eye (36,39) and right eye (42,45) as a simple orientation feature.
# A more robust feature could involve more landmarks or PCA.

orientation_features = []
original_keypoints_list = []
images_list = []

for sample in selected_samples:
    image = sample['image'].numpy() # Image is already normalized and grayscaled by ToTensor
    image = np.transpose(image, (1, 2, 0)) # Change from CxHxW to HxWxC
    image = np.squeeze(image) # Remove channel dimension if it's 1
    images_list.append(image)

    keypoints = sample['keypoints']
    original_keypoints_list.append(un_transform_keypoints(keypoints))

    # Extract x-coordinates for left eye center (average of 36, 39) and right eye center (average of 42, 45)
    # Or a simpler feature: difference between mean x of right-half landmarks and left-half landmarks
    # Let's consider the x-coordinates of the left and right corners of the eyes for a robust orientation feature.
    # Left eye outer corner (36), left eye inner corner (39)
    # Right eye outer corner (45), right eye inner corner (42)
    # Or even simpler: the x-coordinate of nose tip (30) relative to face center, or average x of left/right side

    # Feature: (mean_x_right_eye - mean_x_left_eye) and (mean_x_right_mouth - mean_x_left_mouth)
    # Let's try average x-coordinate of points 0-7 (left jaw) vs 9-16 (right jaw) to indicate head turn
    # Or, the horizontal position of the nose tip (keypoint 30) can indicate left/right turn if normalized to face width.

    # A simple indicator: x-coordinate of the nose tip (keypoint 30)
    # or the difference between the x-coordinates of rightmost and leftmost points (e.g., jawline corners 0 and 16)

    # Let's define orientation based on the x-coordinate of the nose tip (keypoint 30)
    # and the x-coordinates of the outer eye corners (keypoints 36 and 45).
    # A single value representing horizontal shift is enough for 1D clustering.
    nose_tip_x = keypoints[30, 0].item() # x-coordinate of nose tip (normalized)
    orientation_features.append([nose_tip_x])

# Convert to numpy array
orientation_features = np.array(orientation_features)

# Perform K-means clustering
k = 3  # For 'left-oriented', 'right-oriented', and 'centered'
kmeans = KMeans(n_clusters=k, random_state=0, n_init=10)
clusters = kmeans.fit_predict(orientation_features)

# Map clusters to meaningful labels
# Sort centroids to map them to 'left', 'centered', 'right'
sorted_centroid_indices = np.argsort(kmeans.cluster_centers_.flatten())

orientation_map = {
    sorted_centroid_indices[0]: 'right-oriented', # Smallest x means face turned right (our right)
    sorted_centroid_indices[1]: 'centered',
    sorted_centroid_indices[2]: 'left-oriented' # Largest x means face turned left (our left)
}

predicted_orientations = [orientation_map[cluster_id] for cluster_id in clusters]

print("Keypoints extracted and features prepared for clustering.")
print("Predicted orientations for the 5 selected images:", predicted_orientations)

# Store results for visualization in the next step
clustering_results = {
    'images': images_list,
    'keypoints': original_keypoints_list,
    'orientations': predicted_orientations
}

# Display the selected images with ground truth keypoints and predicted orientation
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 6))
for i in range(len(selected_samples)):
    ax = plt.subplot(1, len(selected_samples), i + 1)
    image = clustering_results['images'][i]
    key_pts = clustering_results['keypoints'][i]

    plt.imshow(image, cmap='gray')
    plt.scatter(key_pts[:, 0], key_pts[:, 1], s=20, marker='.', c='g')
    plt.title(f"Orientation: {clustering_results['orientations'][i]}")
    plt.axis('off')
plt.tight_layout()
plt.show()

print("Visualization of selected images with ground truth keypoints and predicted orientations displayed.")
````

## Extract Smiling Features from Keypoints

### Subtask:
Identify specific keypoints related to the mouth (e.g., mouth corners, lip curvature) from the selected samples. Calculate a numerical feature (or features) that can represent whether a face is smiling or not. This might involve distances between keypoints or angles.


**Reasoning**:
I need to initialize a list for smile features, iterate through the selected samples, extract and un-transform keypoints, and then calculate mouth width and height from specific keypoints (48, 54, 51, 57) to represent smile features. These features will then be appended to the `smile_features` list.



````python
import numpy as np

# Initialize an empty list to store the calculated smile features
smile_features = []

# Iterate through each sample in the selected_samples list
for i, sample in enumerate(selected_samples):
    # Extract keypoints (ground truth keypoints)
    keypoints = sample['keypoints']

    # Un-transform the keypoints from normalized to original image coordinates
    untransformed_keypoints = un_transform_keypoints(keypoints)

    # Define keypoints for mouth corners and top/bottom lip for smile feature calculation
    # Keypoint 48: left mouth corner
    # Keypoint 54: right mouth corner
    # Keypoint 51: upper lip top
    # Keypoint 57: lower lip bottom

    # Calculate mouth width (distance between keypoints 48 and 54)
    mouth_width = np.linalg.norm(untransformed_keypoints[48] - untransformed_keypoints[54])

    # Calculate mouth height (distance between keypoints 51 and 57)
    mouth_height = np.linalg.norm(untransformed_keypoints[51] - untransformed_keypoints[57])

    # Append the calculated features (width and height, or a ratio) to the smile_features list
    # For this task, let's append both width and height as a tuple or list
    smile_features.append([mouth_width, mouth_height])

print("Calculated smile features (mouth width, mouth height) for each selected sample:")
for i, features in enumerate(smile_features):
    print(f"Sample {i+1}: Width = {features[0]:.2f}, Height = {features[1]:.2f}")
````

## Interpret and Classify Smiles

### Subtask:
Analyze the centroids of the two clusters to assign meaningful labels: 'smiling' and 'not smiling'. Then, classify each of the selected sample images based on the cluster it belongs to.


**Reasoning**:
I need to analyze the KMeans cluster centroids to assign meaningful labels ('smiling' and 'not smiling') and then classify each sample accordingly. This involves comparing the mouth width and height values in the centroids.



````python
import numpy as np

# Access the cluster centroids
smile_centroids = kmeans_smile.cluster_centers_

print("Cluster Centroids (Mouth Width, Mouth Height):")
for i, centroid in enumerate(smile_centroids):
    print(f"Cluster {i}: Width={centroid[0]:.2f}, Height={centroid[1]:.2f}")

# Determine which centroid corresponds to 'smiling' and 'not smiling'
# A common heuristic is that smiling involves both wider and taller mouths,
# or at least a significantly larger mouth width relative to height.
# Let's use the sum of width and height as a simple indicator.

centroid_scores = [np.sum(c) for c in smile_centroids]

smile_label_map = {}
if centroid_scores[0] > centroid_scores[1]:
    # Cluster 0 has a higher score, so it's 'smiling'
    smile_label_map[0] = 'smiling'
    smile_label_map[1] = 'not smiling'
else:
    # Cluster 1 has a higher score, so it's 'smiling'
    smile_label_map[1] = 'smiling'
    smile_label_map[0] = 'not smiling'

print("\nSmile label mapping:", smile_label_map)

# Classify each sample based on its cluster
predicted_smile_labels = [smile_label_map[cluster_id] for cluster_id in smile_clusters]

print("\nPredicted smile labels for each sample:")
for i, label in enumerate(predicted_smile_labels):
    print(f"Sample {i+1}: {label}")
````

````python
# Redefine the Normalize class to handle color images
import cv2 # Ensure cv2 is imported

class ColorNormalize(object):
    """Normalize a color image (keeping 3 channels) and keypoints."""

    def __call__(self, sample):
        image, key_pts = sample['image'], sample['keypoints']

        image_copy = np.copy(image) # Already HxWxC (RGB) from PIL conversion

        # Scale color range from [0, 255] to [0, 1] for image
        # ToTensor will handle scaling to [0, 1] and permutation to CxHxW
        # So, we just need to ensure the image is a float before ToTensor.
        # If image is already a numpy array from PIL, it might be float, otherwise convert.
        if image_copy.dtype == np.uint8:
            image_copy = image_copy.astype(np.float32) / 255.0

        # scale keypoints to be centered around 0 with a range of [-1, 1]
        key_pts = (key_pts - 100)/50.0

        return {'image': image_copy, 'keypoints': key_pts}

print("ColorNormalize class defined.")
````

I will use the `un_transform_keypoints` function to convert normalized keypoints back to original image coordinates and then calculate mouth width and height from specific keypoints (48, 54, 51, 57) for each sample, storing these as a list of features.



````python
import numpy as np

# Un-transformation function (copied from visualize_output in the notebook context)
def un_transform_keypoints(key_pts_tensor):
    key_pts_np = key_pts_tensor.numpy()
    # undo normalization of keypoints
    key_pts_unnorm = key_pts_np * 50.0 + 100
    return key_pts_unnorm

# Initialize an empty list to store the calculated smile features
smile_features = []

# Iterate through each sample in the selected_samples list
for i, sample in enumerate(selected_samples):
    # Extract keypoints (ground truth keypoints)
    keypoints = sample['keypoints']

    # Un-transform the keypoints from normalized to original image coordinates
    untransformed_keypoints = un_transform_keypoints(keypoints)

    # Define keypoints for mouth corners and top/bottom lip for smile feature calculation
    # Keypoint 48: left mouth corner
    # Keypoint 54: right mouth corner
    # Keypoint 51: upper lip top
    # Keypoint 57: lower lip bottom

    # Calculate mouth width (distance between keypoints 48 and 54)
    mouth_width = np.linalg.norm(untransformed_keypoints[48] - untransformed_keypoints[54])

    # Calculate mouth height (distance between keypoints 51 and 57)
    mouth_height = np.linalg.norm(untransformed_keypoints[51] - untransformed_keypoints[57])

    # Append the calculated features (width and height) to the smile_features list
    smile_features.append([mouth_width, mouth_height])

print("Calculated smile features (mouth width, mouth height) for each selected sample:")
for i, features in enumerate(smile_features):
    print(f"Sample {i+1}: Width = {features[0]:.2f}, Height = {features[1]:.2f}")
````

````python
import numpy as np

# Access the cluster centroids
smile_centroids = kmeans_smile.cluster_centers_

print("Cluster Centroids (Mouth Width, Mouth Height):")
for i, centroid in enumerate(smile_centroids):
    print(f"Cluster {i}: Width={centroid[0]:.2f}, Height={centroid[1]:.2f}")

# Determine which centroid corresponds to 'smiling' and 'not smiling'
# A common heuristic is that smiling involves both wider and taller mouths,
# or at least a significantly larger mouth width relative to height.
# Let's use the sum of width and height as a simple indicator.

centroid_scores = [np.sum(c) for c in smile_centroids]

smile_label_map = {}
if centroid_scores[0] > centroid_scores[1]:
    # Cluster 0 has a higher score, so it's 'smiling'
    smile_label_map[0] = 'smiling'
    smile_label_map[1] = 'not smiling'
else:
    # Cluster 1 has a higher score, so it's 'smiling'
    smile_label_map[1] = 'smiling'
    smile_label_map[0] = 'not smiling'

print("\nSmile label mapping:", smile_label_map)

# Classify each sample based on its cluster
predicted_smile_labels = [smile_label_map[cluster_id] for cluster_id in smile_clusters]

print("\nPredicted smile labels for each sample:")
for i, label in enumerate(predicted_smile_labels):
    print(f"Sample {i+1}: {label}")
````

## Prepare Keypoints for Face Swapping

### Subtask:
Extract the keypoints for both selected images and convert them to a suitable format (e.g., NumPy arrays of integers) for OpenCV's image processing functions. Also, define helper functions for finding the convex hull and Delaunay triangulation of the face region based on keypoints.


**Reasoning**:
I need to extract and un-transform keypoints from the selected samples, convert them to integer NumPy arrays, and define helper functions for convex hull and Delaunay triangulation, as specified in the subtask instructions. This involves reusing the previously defined `un_transform_keypoints` function and adding new functions for OpenCV-compatible operations.



## Implement Face Swapping Logic

### Subtask:
Develop a Python function that encapsulates the face swapping logic. This function will take two images and their keypoints, calculate the affine transformation between corresponding facial regions using the keypoints, warp one face onto the other, and then use seamless cloning to blend the swapped face naturally onto the target image.


````python
import numpy as np
import cv2

# Helper function to un-transform image tensor (defined here for self-containment)
def un_transform_image(img_tensor):
    """Converts a normalized PyTorch image tensor (CxHxW) to an OpenCV-compatible NumPy array (HxW, grayscale, 0-255)."""
    # Assuming img_tensor is (1, H, W) for grayscale
    img_np = img_tensor.squeeze(0).numpy() # Removes the channel dimension (1 -> HxW)
    img_np = (img_np * 255).astype(np.uint8) # Scale to 0-255 and convert to uint8
    return img_np

# The helper functions `un_transform_keypoints`, `calculate_convex_hull`,
# and `calculate_delaunay_triangles` are assumed to be defined in previous cells.
# For the purpose of this isolated block, let's include their definitions as they were
# in cell `8684c0fb` and assume `un_transform_keypoints` takes a PyTorch tensor.

# Re-define `un_transform_keypoints` to ensure it's available
def un_transform_keypoints(key_pts_tensor):
    """Converts normalized keypoints back to original image coordinates."""
    key_pts_np = key_pts_tensor.numpy()
    # undo normalization of keypoints
    key_pts_unnorm = key_pts_np * 50.0 + 100 # Assuming original normalization used 50.0 and 100
    return key_pts_unnorm

# Re-define `calculate_convex_hull` from previous cell `8684c0fb`
def calculate_convex_hull(keypoints_int, image_shape=None): # image_shape is not used in the updated version, but kept for compatibility
    """
    Calculates the convex hull indices of the facial keypoints.
    Args:
        keypoints_int (np.ndarray): Integer array of keypoints (N, 2).
    Returns:
        np.ndarray: Flattened indices of the keypoints forming the convex hull.
    """
    hull_indices = cv2.convexHull(keypoints_int, returnPoints=False)
    return hull_indices.flatten()

# Re-define `calculate_delaunay_triangles` from previous cell `8684c0fb`
def calculate_delaunay_triangles(keypoints_int, image_size):
    """
    Calculates the Delaunay triangulation of keypoints within an image boundary.
    Args:
        keypoints_int (np.ndarray): Integer array of keypoints (N, 2).
        image_size (tuple): Tuple (width, height) defining the bounding rectangle for Delaunay.
    Returns:
        list: A list of triangles, where each triangle is a tuple of 3 keypoint indices.
    """
    rect = (0, 0, image_size[0], image_size[1]) # Bounding rectangle for Delaunay triangulation
    subdiv = cv2.Subdiv2D(rect)

    for p in keypoints_int:
        # Fix: Explicitly cast to int for subdiv.insert
        subdiv.insert((int(p[0]), int(p[1])))

    triangleList = subdiv.getTriangleList()
    delaunay_triangles = []

    for t in triangleList:
        pt1 = (int(t[0]), int(t[1]))
        pt2 = (int(t[2]), int(t[3]))
        pt3 = (int(t[4]), int(t[5]))

        p_indices = []
        for point_to_find in [pt1, pt2, pt3]:
            found_idx = -1
            # Search for the exact point in keypoints_int
            matches = np.where(np.all(keypoints_int == point_to_find, axis=1))[0]
            if len(matches) > 0:
                found_idx = matches[0] # Take the first match
            p_indices.append(found_idx)

        idx1, idx2, idx3 = p_indices[0], p_indices[1], p_indices[2]

        if idx1 != -1 and idx2 != -1 and idx3 != -1 and (idx1 != idx2 and idx1 != idx3 and idx2 != idx3):
            delaunay_triangles.append((idx1, idx2, idx3))

    return delaunay_triangles


def face_swap(src_img_tensor, src_kpts_normalized, dest_img_tensor, dest_kpts_normalized):
    """
    Performs face swapping from a source image to a destination image using facial keypoints.

    Args:
        src_img_tensor (torch.Tensor): The source image tensor (normalized, CxHxW).
        src_kpts_normalized (torch.Tensor): Normalized keypoints for the source face.
        dest_img_tensor (torch.Tensor): The destination image tensor (normalized, CxHxW).
        dest_kpts_normalized (torch.Tensor): Normalized keypoints for the destination face.

    Returns:
        np.ndarray: The resulting image with the swapped face (BGR, uint8).
    """
    # 1. Un-transform images and keypoints
    src_img_untransformed = un_transform_image(src_img_tensor)
    dest_img_untransformed = un_transform_image(dest_img_tensor)

    src_keypoints_int = np.array(un_transform_keypoints(src_kpts_normalized), dtype=int)
    dest_keypoints_int = np.array(un_transform_keypoints(dest_kpts_normalized), dtype=int)

    # Convert images to BGR as seamlessClone and many cv2 ops prefer 3 channels
    src_img_bgr = cv2.cvtColor(src_img_untransformed, cv2.COLOR_GRAY2BGR)
    dest_img_bgr = cv2.cvtColor(dest_img_untransformed, cv2.COLOR_GRAY2BGR)

    # Initialize a blank image for the warped source face (BGR)
    warped_src_face_bgr = np.zeros_like(dest_img_bgr, dtype=np.uint8)

    # Get image dimensions for Delaunay triangulation bounding box
    h, w, _ = dest_img_bgr.shape
    image_size = (w, h)

    # 2. Calculate convex hull of destination face keypoints for seamless cloning mask
    dest_hull_indices = calculate_convex_hull(dest_keypoints_int) # Returns flattened indices
    dest_hull_points = dest_keypoints_int[dest_hull_indices] # Map indices to actual points

    # 3. Calculate Delaunay triangulation for the destination face
    delaunay_triangles = calculate_delaunay_triangles(dest_keypoints_int, image_size)

    # 4. For each triangle in the Delaunay triangulation
    for tri_indices in delaunay_triangles:
        # a. Identify the vertices (keypoints) of the current triangle in both source and destination faces
        pts_src = np.array([src_keypoints_int[idx] for idx in tri_indices], dtype=np.float32)
        pts_dest = np.array([dest_keypoints_int[idx] for idx in tri_indices], dtype=np.float32)

        # Get bounding boxes for the triangles
        rect_src = cv2.boundingRect(pts_src)
        rect_dest = cv2.boundingRect(pts_dest)

        (x_src, y_src, w_src, h_src) = rect_src
        (x_dest, y_dest, w_dest, h_dest) = rect_dest

        # Adjust points for the cropped regions (relative coordinates within bounding box)
        pts_src_rect_shifted = np.array([[pt[0] - x_src, pt[1] - y_src] for pt in pts_src], dtype=np.float32)
        pts_dest_rect_shifted = np.array([[pt[0] - x_dest, pt[1] - y_dest] for pt in pts_dest], dtype=np.float32)

        # b. Compute the affine transformation matrix
        M = cv2.getAffineTransform(pts_src_rect_shifted, pts_dest_rect_shifted)

        # Crop corresponding region from source image (BGR)
        cropped_src_rect_bgr = src_img_bgr[y_src:y_src+h_src, x_src:x_src+w_src]

        # c. Apply this affine transformation to warp the source triangle
        warped_triangle_bgr = cv2.warpAffine(cropped_src_rect_bgr, M, (w_dest, h_dest),
                                             flags=cv2.INTER_LINEAR, borderMode=cv2.BORDER_REFLECT_101)

        # Create mask for the destination triangle within its bounding box
        mask_triangle = np.zeros((h_dest, w_dest), dtype=np.uint8)
        cv2.fillConvexPoly(mask_triangle, pts_dest_rect_shifted.astype(int), 255)

        # Copy the warped triangle to the full warped_src_face_bgr image using the mask
        warped_region_in_full = warped_src_face_bgr[y_dest:y_dest+h_dest, x_dest:x_dest+w_dest]

        # Add the warped triangle to the blank canvas, only where the mask is 255
        warped_src_face_bgr[y_dest:y_dest+h_dest, x_dest:x_dest+w_dest] = cv2.add(
            warped_region_in_full,
            cv2.bitwise_and(warped_triangle_bgr, warped_triangle_bgr, mask=mask_triangle)
        )

    # 5. Create a mask of the destination face using its convex hull.
    mask_seamless = np.zeros_like(dest_img_untransformed, dtype=np.uint8) # Grayscale mask
    cv2.fillConvexPoly(mask_seamless, dest_hull_points.astype(int), 255)

    # Get the center of the mask for seamless cloning
    center_x = int(np.mean(dest_hull_points[:, 0])) # Note: hull_points is (N, 1, 2) from cv2.convexHull, so [:,0] for points
    center_y = int(np.mean(dest_hull_points[:, 1]))
    center_face = (center_x, center_y)

    # 6. Use cv2.seamlessClone to blend the warped source face onto the original destination image
    output_image_bgr = cv2.seamlessClone(
        warped_src_face_bgr,
        dest_img_bgr,
        mask_seamless,
        center_face,
        cv2.NORMAL_CLONE
    )

    # 7. Return the resulting image with the swapped face
    return output_image_bgr

print("The `face_swap` function and its necessary helper functions are defined.")
````

**Reasoning**:
The `face_swap` function and its helper functions have been redefined to resolve the previous error. This will involve re-executing the visualization code block to see the outcome.



````python
import matplotlib.pyplot as plt

# Perform face swap
swapped_face_image = face_swap(
    src_img_tensor=source_sample['image'],
    src_kpts_normalized=source_sample['keypoints'],
    dest_img_tensor=dest_sample['image'],
    dest_kpts_normalized=dest_sample['keypoints']
)

# Un-transform original images for display
original_src_img = un_transform_image(source_sample['image'])
original_dest_img = un_transform_image(dest_sample['image'])

# Display results
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(original_src_img, cmap='gray')
plt.title('Original Source Face')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(original_dest_img, cmap='gray')
plt.title('Original Destination Face')
plt.axis('off')

plt.subplot(1, 3, 3)
# swapped_face_image is already BGR from cv2.seamlessClone, but matplotlib expects RGB
# For grayscale, we can convert it back to grayscale if desired, or just show one channel.
# Given the input was grayscale, the output is likely perceived as grayscale even in BGR.
plt.imshow(cv2.cvtColor(swapped_face_image, cv2.COLOR_BGR2RGB)) # Convert BGR to RGB for correct matplotlib display
plt.title('Face Swapped Result')
plt.axis('off')

plt.tight_layout()
plt.show()

print("Face swapping completed and results displayed.")
````

## Summary:

### Q&A
The entire face swapping process involves several key steps:
1.  **Image and Keypoint Selection**: Two distinct images are selected from a dataset, one as the source and one as the destination, along with their corresponding ground truth facial keypoints.
2.  **Keypoint Preparation**: The normalized keypoints for both images are un-transformed back to original image coordinates and converted into integer NumPy arrays. Helper functions are defined to calculate the convex hull of facial keypoints (to define the face region) and the Delaunay triangulation (to divide the face into smaller, transformable triangles).
3.  **Face Swapping Logic**:
    *   Images and keypoints are un-transformed. Images are converted to BGR format for OpenCV compatibility.
    *   The convex hull of the destination face keypoints is determined to create a mask for seamless cloning.
    *   Delaunay triangulation is applied to the destination face's keypoints, breaking it into many small triangles.
    *   For each corresponding triangle pair between the source and destination faces, an affine transformation matrix is computed.
    *   The corresponding source triangle region is warped using this affine transformation to fit the destination triangle's shape.
    *   The warped source triangles are assembled onto a blank canvas, forming a warped source face.
    *   Finally, OpenCV's `seamlessClone` function is used to blend this warped source face onto the original destination image using the destination face's convex hull as a mask, ensuring a natural blend.

The role of keypoints is fundamental: they serve as anchor points to precisely map facial features between the source and destination faces. By using these keypoints to define triangles and calculate affine transformations, the algorithm can accurately warp facial regions from one image to another.

The effectiveness of the swapping and blending technique relies on:
*   **Accurate Keypoint Detection**: The quality of the swap is highly dependent on the precision of the facial keypoints.
*   **Delaunay Triangulation**: This technique allows for localized warping, preserving facial features' relative positions within triangles.
*   **Affine Transformations**: These transformations ensure that the warped facial regions maintain their internal structure while adapting to the destination face's geometry.
*   **Seamless Cloning**: This final blending step is crucial for making the swapped face appear natural by smoothing transitions at the boundaries of the inserted face.
