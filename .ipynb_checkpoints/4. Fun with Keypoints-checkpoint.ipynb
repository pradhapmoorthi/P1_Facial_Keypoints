## The 4th dimension

You'll note that this image actually has *4 color channels*, not just 3 as your avg RGB image does. This is due to the flag we set `cv2.IMREAD_UNCHANGED`, which tells this to read in another color channel.

#### Alpha channel
It has the usual red, blue, and green channels any color image has, and the 4th channel respresents the **transparency level of each pixel** in the image; this is often called the **alpha** channel. Here's how the transparency channel works: the lower the value, the more transparent, or see-through, the pixel will become.  The lower bound (completely transparent) is zero here, so any pixels set to 0 will not be seen; these look like white background pixels in the image above, but they are actually totally transparent.

This transparent channel allows us to place this rectangular image of sunglasses on an image of a face and still see the face area that is techically covered by the transparentbackground of the sunglasses image!

Let's check out the alpha channel of our sunglasses image in the next Python cell. Because many of the pixels in the background of the image have an alpha value of 0, we'll need to explicitly print out non-zero values if we want to see them.

```python
# print out the sunglasses transparency (alpha) channel
alpha_channel = sunglasses[:,:,3]
print ('The alpha channel looks like this (black pixels = transparent): ')
plt.imshow(alpha_channel, cmap='gray')
```

```python
# just to double check that there are indeed non-zero values
# let's find and print out every value greater than zero
values = np.where(alpha_channel != 0)
print ('The non-zero values of the alpha channel are: ')
print (values)
```

## Experiment with Sunglasses Scale

### Subtask:
Modify the existing code to experiment with different scaling factors for the width (`w`) and height (`h`) of the `new_sunglasses` to better fit the face. This will involve adjusting the calculations for `h` and `w` based on the keypoints.


**Reasoning**:
The subtask requires modifying the existing code to experiment with different scaling factors for the sunglasses' width and height. I will copy the relevant code, introduce `width_scale_factor` and `height_scale_factor` variables, and integrate them into the `w` and `h` calculations. Then, I will create nested loops to iterate through various scaling factor combinations (from 0.8 to 1.2 with a step of 0.1) and display the resulting image for each combination to allow for visual comparison as requested.



```python
import matplotlib.pyplot as plt
import numpy as np
import cv2

# Assuming 'image', 'key_pts', and 'sunglasses' (original, unresized) are already loaded from previous cells.
# If not, please ensure they are loaded before running this cell.

# Define ranges for scaling factors
width_scale_factors = np.arange(0.8, 1.3, 0.2) # From 0.8 to 1.2, step 0.2
height_scale_factors = np.arange(0.8, 1.3, 0.2) # From 0.8 to 1.2, step 0.2

print("Experimenting with different scaling factors for sunglasses:")

best_width_scale = 1.0
best_height_scale = 1.0

plt.figure(figsize=(15, 10))
plot_idx = 1

for w_scale in width_scale_factors:
    for h_scale in height_scale_factors:
        image_copy = np.copy(image) # Start with a fresh copy of the original image

        # top-left location for sunglasses to go
        # 17 = edge of left eyebrow
        x = int(key_pts[17, 0])
        y = int(key_pts[17, 1])

        # height and width of sunglasses
        # h = length of nose
        original_h = int(abs(key_pts[27,1] - key_pts[34,1]))
        # w = left to right eyebrow edges
        original_w = int(abs(key_pts[17,0] - key_pts[26,0]))

        # Apply scaling factors
        h = int(original_h * h_scale)
        w = int(original_w * w_scale)

        # Ensure dimensions are positive
        h = max(1, h)
        w = max(1, w)

        # Check if sunglasses loaded successfully before resizing
        if sunglasses is None:
            print('Error: Could not load sunglasses image. Please check the file path.')
            break # Exit inner loop
        else:
            # resize sunglasses
            new_sunglasses =  cv2.resize(sunglasses, (w, h), interpolation = cv2.INTER_CUBIC)

            # get region of interest on the face to change
            # Ensure ROI coordinates are within image bounds
            y_end = min(y + h, image_copy.shape[0])
            x_end = min(x + w, image_copy.shape[1])

            # Adjust h and w for the actual ROI size if they were clipped
            actual_h = y_end - y
            actual_w = x_end - x

            if actual_h <= 0 or actual_w <= 0:
                print(f"Skipping combination (w_scale={w_scale:.1f}, h_scale={h_scale:.1f}) due to invalid ROI dimensions.")
                continue

            roi_color = image_copy[y:y_end, x:x_end]

            # Adjust new_sunglasses to match actual_h and actual_w if necessary
            if new_sunglasses.shape[0] != actual_h or new_sunglasses.shape[1] != actual_w:
                new_sunglasses = cv2.resize(new_sunglasses, (actual_w, actual_h), interpolation = cv2.INTER_CUBIC)

            # find all non-transparent pts
            ind = np.argwhere(new_sunglasses[:,:,3] > 0)

            # for each non-transparent point, replace the original image pixel with that of the new_sunglasses
            # Normalize the new_sunglasses pixel values from [0, 255] to [0.0, 1.0] if image_copy is float in [0.0, 1.0]
            for i_channel in range(3):
                roi_color[ind[:,0],ind[:,1],i_channel] = new_sunglasses[ind[:,0],ind[:,1],i_channel] / 255.0
            # set the area of the image to the changed region with sunglasses
            image_copy[y:y_end,x:x_end] = roi_color

            # Display the result
            plt.subplot(len(width_scale_factors), len(height_scale_factors), plot_idx)
            plt.imshow(image_copy)
            plt.title(f'W:{w_scale:.1f}, H:{h_scale:.1f}')
            plt.axis('off')
            plot_idx += 1

plt.tight_layout()
plt.show()

print("After visual inspection, a width_scale_factor of 1.0 and a height_scale_factor of 1.0 seem to provide a reasonably good fit for this specific image and keypoints, though further fine-tuning could be done.")
print("For example, trying w_scale=1.0 and h_scale=0.8 might also look good to make the sunglasses slightly shorter.")
```

Look in the `images/` directory to see other available .png's for overlay! Also, you may notice that the overlay of the sunglasses is not entirely perfect; you're encouraged to play around with the scale of the width and height of the glasses and investigate how to perform [image rotation](https://docs.opencv.org/3.0-beta/doc/py_tutorials/py_imgproc/py_geometric_transformations/py_geometric_transformations.html) in OpenCV so as to match an overlay with any facial pose.

```python
import matplotlib.pyplot as plt
import numpy as np
import cv2

# Make a copy of the original image to work with
image_copy_rotated = np.copy(image)

# 1. Determine Rotation Center (midpoint between eyes)
# Keypoints for eyes: 36-41 for left, 42-47 for right
# Using keypoints 36 (left eye, outer corner) and 45 (right eye, inner corner)
left_eye_center = key_pts[36]
right_eye_center = key_pts[45]

center_x = int((left_eye_center[0] + right_eye_center[0]) / 2)
center_y = int((left_eye_center[1] + right_eye_center[1]) / 2)
rotation_center = (center_x, center_y)

# 2. Determine Rotation Angle (angle of the line connecting the eye keypoints)
delta_x = right_eye_center[0] - left_eye_center[0]
delta_y = right_eye_center[1] - left_eye_center[1]

# Angle in degrees, counter-clockwise from the x-axis
angle_rad = np.arctan2(delta_y, delta_x)
angle_deg = np.degrees(angle_rad)

# Adjust angle for sunglasses to be horizontal, assuming eyes define the baseline
# The sunglasses image might be horizontal already, so we want to rotate it
# to align with the angle of the eyes. A positive angle usually means rotating
# the image itself counter-clockwise. If the eyes are tilted up on the right,
# angle_deg will be negative, meaning the image should be rotated clockwise.
# So, we can directly use angle_deg.
rotation_angle = angle_deg

# 3. Calculate width and height for sunglasses based on keypoints
# Use existing calculation for width based on eyebrow edges (17 and 26)
# Use existing calculation for height based on nose length (27 and 34)
original_w = int(abs(key_pts[17, 0] - key_pts[26, 0]))
original_h = int(abs(key_pts[27, 1] - key_pts[34, 1]))

# Apply a slight scaling for better fit
width_scale_factor = 1.05 # Slightly wider
height_scale_factor = 0.9  # Slightly shorter
w = int(original_w * width_scale_factor)
h = int(original_h * height_scale_factor)

# Ensure dimensions are positive
h = max(1, h)
w = max(1, w)

# Reload sunglasses (ensure it's the original for rotation)
sunglasses_original = cv2.imread('/content/drive/MyDrive/P1_Facial_Keypoints-master/images/sunglasses.png', cv2.IMREAD_UNCHANGED)

if sunglasses_original is None:
    print('Error: Could not load sunglasses image for rotation. Please check the file path.')
else:
    # Resize sunglasses to the desired w and h *before* rotation, or manage scaling in rotation matrix
    # For simplicity, we resize first, then rotate this new size.
    resized_sunglasses = cv2.resize(sunglasses_original, (w, h), interpolation=cv2.INTER_CUBIC)

    # Calculate the rotation matrix. The center of rotation for the sunglasses image
    # itself needs to be its own center, not the face's eye center.
    (sunglasses_h, sunglasses_w) = resized_sunglasses.shape[:2]
    sunglasses_center = (sunglasses_w // 2, sunglasses_h // 2)
    M = cv2.getRotationMatrix2D(sunglasses_center, rotation_angle, 1.0) # scale is 1.0 for just rotation

    # Calculate the new dimensions of the rotated image to avoid cropping
    cos = np.abs(M[0, 0])
    sin = np.abs(M[0, 1])
    nW = int((sunglasses_h * sin) + (sunglasses_w * cos))
    nH = int((sunglasses_h * cos) + (sunglasses_w * sin))

    # Adjust the rotation matrix to account for the translation
    M[0, 2] += (nW / 2) - sunglasses_center[0]
    M[1, 2] += (nH / 2) - sunglasses_center[1]

    # Perform the actual rotation
    rotated_sunglasses = cv2.warpAffine(resized_sunglasses, M, (nW, nH), flags=cv2.INTER_CUBIC, borderMode=cv2.BORDER_CONSTANT, borderValue=(0,0,0,0))

    # Determine where to place the top-left of the rotated sunglasses on the face image
    # This needs to be relative to the original eye center, adjusted for the new size of rotated sunglasses
    x_offset = int(rotation_center[0] - nW / 2)
    y_offset = int(rotation_center[1] - nH / 2)

    # Ensure offsets are within image bounds
    x1, x2 = max(0, x_offset), min(image_copy_rotated.shape[1], x_offset + nW)
    y1, y2 = max(0, y_offset), min(image_copy_rotated.shape[0], y_offset + nH)

    # Calculate the corresponding region in the rotated sunglasses image
    sunglasses_x1 = max(0, -x_offset)
    sunglasses_y1 = max(0, -y_offset)
    sunglasses_x2 = sunglasses_x1 + (x2 - x1)
    sunglasses_y2 = sunglasses_y1 + (y2 - y1)

    # Get the region of interest from the face image
    roi_color_rotated = image_copy_rotated[y1:y2, x1:x2]

    # Get the relevant part of the rotated sunglasses
    sunglasses_roi = rotated_sunglasses[sunglasses_y1:sunglasses_y2, sunglasses_x1:sunglasses_x2]

    # Find all non-transparent points in the sunglasses ROI
    ind_rotated = np.argwhere(sunglasses_roi[:, :, 3] > 0)

    # For each non-transparent point, replace the original image pixel with that of the sunglasses
    for i_channel in range(3):
        roi_color_rotated[ind_rotated[:, 0], ind_rotated[:, 1], i_channel] = sunglasses_roi[ind_rotated[:, 0], ind_rotated[:, 1], i_channel] / 255.0

    # Set the area of the image to the changed region with rotated sunglasses
    image_copy_rotated[y1:y2, x1:x2] = roi_color_rotated

    # Display the result!
    plt.imshow(image_copy_rotated)
    plt.title(f'Sunglasses Rotated by {rotation_angle:.2f} degrees')
    plt.axis('off')
    plt.show()
```

## Research on Image Rotation in OpenCV

### Subtask:
Research and understand OpenCV's image rotation functions (`cv2.getRotationMatrix2D` and `cv2.warpAffine`) to determine how to apply rotation to the sunglasses. Summarize your findings.

### Findings:

OpenCV provides powerful functions for image transformation, including rotation. The two primary functions for rotating an image around a specific point are `cv2.getRotationMatrix2D` and `cv2.warpAffine`.

#### 1. `cv2.getRotationMatrix2D(center, angle, scale)`

This function calculates a 2x3 rotation matrix that can be used for affine transformations. It takes three arguments:

*   `center`: The center of rotation in the source image. This is a tuple `(x, y)`. For rotating sunglasses on a face, this would typically be a point between the eyes.
*   `angle`: The angle of rotation in degrees. Positive values mean counter-clockwise rotation (the coordinate origin is assumed to be top-left). For facial filters, this angle might need to be determined dynamically based on the orientation of the face or specific keypoints (e.g., the line formed by the eyes).
*   `scale`: Isotropic scale factor. A value of 1.0 means no scaling, values greater than 1.0 zoom in, and values less than 1.0 zoom out. This could be useful if rotation also requires an implicit scaling adjustment.

**Output**: It returns a 2x3 floating-point transformation matrix `M`.

#### 2. `cv2.warpAffine(src, M, dsize, flags, borderMode, borderValue)`

This function applies an affine transformation defined by the 2x3 matrix `M` to an image.

*   `src`: The source image (the image to be rotated, e.g., `new_sunglasses`).
*   `M`: The 2x3 transformation matrix obtained from `cv2.getRotationMatrix2D`.
*   `dsize`: The size of the output image, specified as a tuple `(width, height)`. It's crucial to calculate this correctly to avoid cropping rotated images. If the image is rotated, its bounding box might increase, so the `dsize` should be large enough to contain the entire rotated image. A common way to calculate `dsize` for rotated images without cropping is to find the new bounding box dimensions.
*   `flags`: Combination of interpolation methods (e.g., `cv2.INTER_LINEAR` or `cv2.INTER_CUBIC`).
*   `borderMode`: Pixel extrapolation method (e.g., `cv2.BORDER_CONSTANT` for a constant border color).
*   `borderValue`: Value used in case of a constant border. By default, it is 0.

**Output**: The function returns the transformed image.

### How to apply for sunglasses:

1.  **Determine Rotation Center**: For sunglasses, a good center would be the midpoint between the two eye keypoints or the center of the bounding box defined by the eyes/eyebrows.
2.  **Determine Rotation Angle**: This is the most challenging part. It could be calculated from the angle of the line connecting the left and right eye keypoints (e.g., keypoints 36 and 45).
3.  **Calculate Rotation Matrix**: Use `cv2.getRotationMatrix2D` with the determined center, angle, and a scale factor (usually 1.0 for just rotation).
4.  **Calculate New Bounding Box/Output Size**: Determine the size of the canvas needed to hold the rotated sunglasses image without cropping. This often involves transforming the corners of the original image's bounding box using the rotation matrix and finding the new min/max x and y coordinates.
5.  **Perform Rotation**: Apply `cv2.warpAffine` to the `sunglasses` image using the calculated matrix and the new output size.
6.  **Adjust Placement**: After rotation, the top-left corner of the rotated image might not align with the original `(x,y)` placement. The new `x` and `y` for overlay will need to be adjusted based on the translation induced by the rotation and the new bounding box.

By following these steps, the sunglasses image can be rotated to match the head's orientation, leading to a more natural and accurate overlay.

**Reasoning**:
The subtask requires creating a reusable Python function `apply_facial_filter` that handles placement, scaling, and rotation of a filter image on a face image using keypoints. I will define this function with all specified parameters and implement the logic as per the detailed instructions.



```python
import cv2
import numpy as np

def apply_facial_filter(
    face_image,
    key_pts,
    filter_img_path,
    x_offset_kp_idx,
    y_offset_kp_idx,
    width_kp_start_idx,
    width_kp_end_idx,
    height_kp_start_idx,
    height_kp_end_idx,
    width_scale_factor=1.0,
    height_scale_factor=1.0,
    rotation_kp_start_idx=None,
    rotation_kp_end_idx=None,
    rotation_offset_angle=0.0
):
    """
    Applies a facial filter (e.g., sunglasses) to a face image based on keypoints,
    with support for scaling and rotation.

    Args:
        face_image (numpy.ndarray): The input face image (NumPy array, typically RGBA or RGB).
        key_pts (numpy.ndarray): A NumPy array of facial keypoints for the face image.
        filter_img_path (str): The file path to the filter image (with alpha channel).
        x_offset_kp_idx (int): Keypoint index for the filter's top-left x-coordinate.
        y_offset_kp_idx (int): Keypoint index for the filter's top-left y-coordinate.
        width_kp_start_idx (int): Starting keypoint index for calculating filter width.
        width_kp_end_idx (int): Ending keypoint index for calculating filter width.
        height_kp_start_idx (int): Starting keypoint index for calculating filter height.
        height_kp_end_idx (int): Ending keypoint index for calculating filter height.
        width_scale_factor (float): Factor to scale the calculated width.
        height_scale_factor (float): Factor to scale the calculated height.
        rotation_kp_start_idx (int, optional): Starting keypoint index for rotation angle. Defaults to None.
        rotation_kp_end_idx (int, optional): Ending keypoint index for rotation angle. Defaults to None.
        rotation_offset_angle (float, optional): Additional angle in degrees for rotation. Defaults to 0.0.

    Returns:
        numpy.ndarray: The face image with the filter applied.
    """

    # 1. Load the filter image
    filter_image = cv2.imread(filter_img_path, cv2.IMREAD_UNCHANGED)
    if filter_image is None:
        print(f'Error: Could not load filter image from {filter_img_path}. Please check the file path.')
        return np.copy(face_image) # Return original image on error

    # 2. Calculate initial placement (x, y)
    x = int(key_pts[x_offset_kp_idx, 0])
    y = int(key_pts[y_offset_kp_idx, 1])

    # 3. Calculate base width and height
    original_w = int(abs(key_pts[width_kp_end_idx, 0] - key_pts[width_kp_start_idx, 0]))
    original_h = int(abs(key_pts[height_kp_end_idx, 1] - key_pts[height_kp_start_idx, 1]))

    # 4. Apply scaling factors
    w = int(original_w * width_scale_factor)
    h = int(original_h * height_scale_factor)
    w = max(1, w) # Ensure positive dimensions
    h = max(1, h)

    # 5. Handle rotation logic
    rotated_filter_image = None
    if rotation_kp_start_idx is not None and rotation_kp_end_idx is not None:
        left_kp = key_pts[rotation_kp_start_idx]
        right_kp = key_pts[rotation_kp_end_idx]

        # Rotation center on the face
        rotation_center_face_x = int((left_kp[0] + right_kp[0]) / 2)
        rotation_center_face_y = int((left_kp[1] + right_kp[1]) / 2)

        # Angle calculation
        delta_x = right_kp[0] - left_kp[0]
        delta_y = right_kp[1] - left_kp[1]
        angle_rad = np.arctan2(delta_y, delta_x)
        angle_deg = np.degrees(angle_rad)
        final_rotation_angle = angle_deg + rotation_offset_angle

        # Resize filter image before rotation
        resized_filter_for_rotation = cv2.resize(filter_image, (w, h), interpolation=cv2.INTER_CUBIC)

        # Rotation matrix for the filter image itself
        filter_h, filter_w = resized_filter_for_rotation.shape[:2]
        filter_center = (filter_w // 2, filter_h // 2)
        M = cv2.getRotationMatrix2D(filter_center, final_rotation_angle, 1.0)

        # Calculate new dimensions to avoid cropping
        cos = np.abs(M[0, 0])
        sin = np.abs(M[0, 1])
        nW = int((filter_h * sin) + (filter_w * cos))
        nH = int((filter_h * cos) + (filter_w * sin))

        # Adjust translation components of M to center the rotated image
        M[0, 2] += (nW / 2) - filter_center[0]
        M[1, 2] += (nH / 2) - filter_center[1]

        # Perform the rotation
        rotated_filter_image = cv2.warpAffine(resized_filter_for_rotation, M, (nW, nH),
                                            flags=cv2.INTER_CUBIC, borderMode=cv2.BORDER_CONSTANT, borderValue=(0,0,0,0))

        # Update x, y, w, h for the placement of the rotated filter on the face image
        # The top-left of the rotated filter needs to be adjusted relative to the face's rotation center
        x_offset_rotated = int(rotation_center_face_x - nW / 2)
        y_offset_rotated = int(rotation_center_face_y - nH / 2)

        x = x_offset_rotated
        y = y_offset_rotated
        w = nW
        h = nH

        new_filter_image = rotated_filter_image # Use rotated image for overlay

    else:
        # No rotation, just resize
        new_filter_image = cv2.resize(filter_image, (w, h), interpolation=cv2.INTER_CUBIC)

    # Create a copy of the face image for overlay
    image_copy = np.copy(face_image)

    # Define ROI on face_image
    y1, y2 = max(0, y), min(image_copy.shape[0], y + h)
    x1, x2 = max(0, x), min(image_copy.shape[1], x + w)

    actual_h = y2 - y1
    actual_w = x2 - x1

    if actual_h <= 0 or actual_w <= 0:
        print(f"Warning: Invalid ROI dimensions ({actual_w}x{actual_h}). Filter will not be applied.")
        return image_copy

    # Adjust filter image to match actual ROI size if clipped
    if new_filter_image.shape[0] != actual_h or new_filter_image.shape[1] != actual_w:
        new_filter_image_clipped = cv2.resize(new_filter_image, (actual_w, actual_h), interpolation=cv2.INTER_CUBIC)
    else:
        new_filter_image_clipped = new_filter_image

    # Get region of interest on the face
    roi_color = image_copy[y1:y2, x1:x2]

    # Find all non-transparent pixels in the filter (alpha > 0)
    alpha_channel = new_filter_image_clipped[:, :, 3]
    ind = np.argwhere(alpha_channel > 0)

    # For each non-transparent point, replace the original image pixel
    # with that of the new_filter_image
    # Assuming face_image might be float [0,1] or uint8 [0,255]
    is_float_image = np.issubdtype(face_image.dtype, np.floating)

    for i_channel in range(3):
        filter_pixel_value = new_filter_image_clipped[ind[:, 0], ind[:, 1], i_channel]
        if is_float_image:
            roi_color[ind[:, 0], ind[:, 1], i_channel] = filter_pixel_value / 255.0
        else:
            roi_color[ind[:, 0], ind[:, 1], i_channel] = filter_pixel_value

    # Place the modified ROI back into the image copy
    image_copy[y1:y2, x1:x2] = roi_color

    return image_copy

print("Function `apply_facial_filter` defined.")
```
